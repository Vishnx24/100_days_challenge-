TYPE HINT + DATA CLASS

Problem statement

Create a Python program that stores configuration settings for multiple AI models in structured objects.

Your program should support:

Model name

Provider

Temperature

Maximum tokens

System prompt

Printing configuration details

```python
from dataclasses import dataclass

@dataclass
class AIconfig:
    model_name:str
    provider:str
    temperature:float
    maximum_token:int
    system_prompt:str

chat_config=AIconfig(
    model_name="gpt-4o-mini",
    provider="openai",
    temperature=0.7,
    maximum_token=2000,
    system_prompt="You are a helpful assistant."
)
code_config=AIconfig(
    
    model_name="gpt-4o-mini",
    provider="openai",
    temperature=0.2,
    maximum_token=2000,
    system_prompt="You are a helpful assistant that writes code."
)

print(chat_config)
print(code_config)
```
