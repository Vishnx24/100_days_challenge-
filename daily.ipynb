Iterators & Generators

```python
a=[10,34,56,23,56]
it=iter(a)

print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
 # This will raise StopIteration since there are no more elements in the iterator.
```

**generator**

```python
def generator_no():
    yield 1
    yield 2
    yield 2
    yield 3
for i in generator_no():
    print (i)
```

```python
def data_batches(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]


records = list(range(1, 16))

for batch in data_batches(records, 4):
    print(batch)
```
