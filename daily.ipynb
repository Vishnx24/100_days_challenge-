Iterators & Generators

```python
a=[10,34,56,23,56]
it=iter(a)

print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
 # This will raise StopIteration since there are no more elements in the iterator.
```

**generator**

```python
def generator_no():
    yield 1
    yield 2
    yield 2
    yield 3
for i in generator_no():
    print (i)
```
